```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates # X축 날짜 서식을 위한 라이브러리

# 자신의 운영체제에 맞는 폰트 이름을 설정해주세요.
try:
    plt.rc('font', family='Malgun Gothic') 
except:
    print("Malgun Gothic 폰트가 없어 기본 폰트로 설정됩니다.")
    pass

try:
    df_wide = pd.read_excel('매출 데이터/참치캔 총매출.xlsx')
    df_melt = df_wide.melt(id_vars=['구분'], var_name='날짜_str', value_name='매출')

    # '2020년 01월' 같은 텍스트를 실제 날짜 형식으로 변환합니다.
    df_melt['날짜'] = pd.to_datetime(df_melt['날짜_str'].str.replace('년 ', '-').str.replace('월', ''))
    df_melt['매출'] = pd.to_numeric(df_melt['매출'].astype(str).str.replace(',', ''))
    df_final = df_melt[['날짜', '매출']].sort_values('날짜')
    
    print(df_final.head())

    # 3. 그래프 데이터 선택
    x_data = df_final['날짜']
    y_data = df_final['매출']

    # 4. 그래프 생성 및 꾸미기
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.plot(x_data, y_data, marker='o', linestyle='-', label='동원 F&B')

    # 5. X축 눈금 서식 지정 (날짜 형식으로 바뀐 데이터에 적용)
    # locator: 3, 6, 9, 12월 위치에만 눈금을 찍도록 설정
    ax.xaxis.set_major_locator(mdates.MonthLocator(bymonth=[3, 6, 9, 12]))
    # formatter: 눈금 글씨를 '2020-03' 같은 형식으로 설정
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))

    ax.set_title('참치캔 동원 F&B 추이')
    ax.set_xlabel('날짜')
    ax.set_ylabel('총매출')
    ax.grid(True)
    ax.legend()
    fig.autofmt_xdate() # x축 라벨이 겹치지 않게 자동 정렬
    plt.tight_layout()

    # 6. 그래프 보여주기
    plt.show()
    
except FileNotFoundError:
    print("오류: 파일을 찾을 수 없습니다.")
except Exception as e:
    print(f"오류 발생: {e}")
```
