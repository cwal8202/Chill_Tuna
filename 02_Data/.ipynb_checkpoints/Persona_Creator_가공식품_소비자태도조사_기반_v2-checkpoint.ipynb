### 번외 - 설문조사 질문 임베딩 및 유사도검사 시도 코드

```python
import pandas as pd
import numpy as np
from openai import OpenAI
from sklearn.metrics.pairwise import cosine_similarity
from dotenv import load_dotenv
load_dotenv()

# -------------------------
# 1. Upstage Embedding 모델 초기화
# -------------------------
client = OpenAI(api_key="UPSTAGE_API_KEY",  # 🔑 Upstage API Key 넣기
                base_url="https://api.upstage.ai/v1")

EMBED_MODEL = "embedding-query" 

# -------------------------
# 2. 카테고리 정의 + 대표 문장
# -------------------------
category_prompts = {
    "innovation_adoption": "새로운 제품과 혁신을 빠르게 받아들이는 소비 성향",
    "price_sensitivity": "가격에 민감하게 반응하고 할인이나 가성비를 중요시하는 성향",
    "health_orientation": "건강, 영양, 저칼로리, 유기농 등을 중시하는 성향",
    "premium_orientation": "프리미엄, 고급, 품질을 중시하는 소비 성향",
    "cooking_convenience": "조리 편의성과 간편함을 중시하는 성향",
    "brand_loyalty": "특정 브랜드를 선호하고 충성도가 높은 성향",
    "hmr_preference": "가정간편식(HMR)과 즉석식품을 선호하는 성향",
    "variety_seeking": "다양한 제품을 경험하고 새로운 것을 찾는 성향",
}

# -------------------------
# 3. 카테고리 임베딩 생성
# -------------------------
def get_embedding(text):
    response = client.embeddings.create(
        model=EMBED_MODEL,
        input=text
    )
    return response.data[0].embedding

category_embeddings = {cat: get_embedding(desc) for cat, desc in category_prompts.items()}

# -------------------------
# 4. 데이터 불러오기
# -------------------------
excel_file = "segment/00_2024년 가공식품 소비자태도조사_가구용.xlsx"
df = pd.read_excel(excel_file, sheet_name="Numeric")

# 질문 컬럼 추출 (예: Q1, Q2 …)
question_cols = [col for col in df.columns if col.startswith("Q")]

# -------------------------
# 5. 질문 → 카테고리 매핑
# -------------------------
def map_question_to_category(question_text):
    q_emb = get_embedding(question_text)
    sims = {cat: cosine_similarity([q_emb], [emb])[0][0] for cat, emb in category_embeddings.items()}
    best_cat = max(sims, key=sims.get)
    return best_cat, sims

# -------------------------
# 6. 실행 예시
# -------------------------
sample_questions = question_cols[:5]  # 앞 5개만 샘플
for q in sample_questions:
    cat, sims = map_question_to_category(q)
    print(f"{q} → {cat} (Top1), 세부 점수: {sims}")
```

### 응답에 대한 점수 매핑 + 정규화

```python
# 4. 응답 → 점수 매핑
## long-format으로 변환 (ID, Question, Answer)
df_long = df_filtered.melt(id_vars=id_col, var_name="Question", value_name="Answer")

# 매핑 테이블과 merge
df_scored = df_long.merge(df_mapping[["Question", "Answer", "categories", "score"]],
                          on=["Question", "Answer"],
                          how="left")

# 5. 멀티태깅 처리
split_mode = "duplicate"  # "distribute"=점수 분할, "duplicate"=동일 점수 부여

def expand_categories(row):
    if pd.isna(row["categories"]):
        return []
    # 문자열 정리
    cats_str = str(row["categories"]).replace("[", "").replace("]", "")
    cats = cats_str.split(",")
    score = row["score"]
    if split_mode == "distribute":
        score_each = score / len(cats)
        return [(c.strip().replace, score_each) for c in cats]
    else:  # duplicate
        return [(c.strip(), score) for c in cats]

expanded_rows = []
for _, row in df_scored.iterrows():
    for cat, s in expand_categories(row):
        expanded_rows.append([row[id_col], cat, s])

df_expanded = pd.DataFrame(expanded_rows, columns=[id_col, "categories", "score"])

# 6. 카테고리별 점수 집계
df_category_score = (
    df_expanded
    .groupby([id_col, "categories"])["score"]
    .sum()
    .reset_index()
    .pivot(index=id_col, columns="categories", values="score")
    .fillna(0)
    .reset_index()
)


## 6-1. 카테고리별 만점 계산 (정규화용)
expanded_for_max = []
for _, row in df_mapping.iterrows():
    for cat, s in expand_categories(row):
        expanded_for_max.append([cat, s])

df_mapping_expanded = pd.DataFrame(expanded_for_max, columns=["categories", "score"])
max_scores = df_mapping_expanded.groupby("categories")["score"].sum()

## 6-2. 정규화 (0~1 스케일)
df_category_score_norm = df_category_score.copy()
for cat in max_scores.index:
    if cat in df_category_score_norm.columns:
        df_category_score_norm[cat] = df_category_score_norm[cat] / max_scores[cat]


# 7. 인구통계 데이터와 병합
df_result = df_demo.merge(df_category_score_norm, on=id_col, how="left")


# 8. 저장
out_file = "segment/01_category_scores_with_demo_multitag_normalized.csv"
df_result.to_csv(out_file, index=False, encoding="cp949")

print("멀티태깅 + 정규화 반영 최종 데이터 저장 완료:", out_file)
df_result.head()
```

# 2. K-means 군집분석
## 1) 실루엣 계수를 이용한 최적 군집수 탐색
### 스케일 방식에 따른 실루엣 계수 확인

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler, MinMaxScaler, QuantileTransformer
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

file_path = "segment/01_category_scores_with_demo_multitag_normalized.csv"
df = pd.read_csv(file_path, encoding="cp949")

scalers = {
    "StandardScaler": StandardScaler(),
    "MinMaxScaler": MinMaxScaler(),
    "QuantileTransformer": QuantileTransformer(output_distribution="normal", random_state=42)
}

results = {}
start = 3

# 범주형/수치형 구분
cat_vars = demo_cols
num_vars = [c for c in df.columns if c not in demo_cols + [id_col]]

for name, scaler in scalers.items():
    # 수치형 변수만 스케일링
    X_scaled = df.copy()
    X_scaled[num_vars] = scaler.fit_transform(X_scaled[num_vars])
    
    # k=3~10 클러스터링 & 실루엣 계산
    silhouette_scores = []
    for k in range(start, 11):
        km = KMeans(n_clusters=k, random_state=42, n_init=100)
        labels = km.fit_predict(X_scaled[num_vars])
        score = silhouette_score(X_scaled[num_vars], labels)
        silhouette_scores.append(score)
    results[name] = silhouette_scores
    print(f"{name} best silhouette: {max(silhouette_scores):.4f} (k={silhouette_scores.index(max(silhouette_scores))+start})")

# 그래프 비교
plt.figure(figsize=(8,5))
for name, scores in results.items():
    plt.plot(range(start, 11), scores, marker="o", label=name)
plt.xlabel("Number of clusters (k)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score by Scaler")
plt.legend()
plt.show()
```

## 2) K-means 군집 분석

### 수치형(MinMax) 만 적용

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, OneHotEncoder, MinMaxScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

file_path = "segment/01_category_scores_with_demo_multitag_normalized.csv"
df = pd.read_csv(file_path, encoding="cp949")

# 1. 입력 데이터 (df_result = 인구통계 + 카테고리 점수 데이터)

id_col = "id"
demo_cols = ["SQ3", "HQ2","SQ4", "SQ5", "SQ8", "DQ1", "DQ3", "DQ3_1", "DQ4"]

# 범주형/수치형 구분
cat_vars = demo_cols
num_vars = [c for c in df.columns if c not in demo_cols + [id_col]]

X = df[num_vars]

# -------------------------------------
# 2. 전처리 파이프라인
# -------------------------------------
preprocessor = MinMaxScaler()


# -------------------------------------
# 3. 최적 클러스터 개수 탐색 (실루엣 계수)
# -------------------------------------
silhouette_scores = {}
for k in range(3, 9):  # 3~8개 클러스터 탐색
    kmeans_pipe = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("cluster", KMeans(n_clusters=k, random_state=42, n_init=200))
    ])
    
    cluster_labels = kmeans_pipe.fit_predict(X)
    
    # 실루엣 점수 계산
    score = silhouette_score(
        kmeans_pipe.named_steps["preprocessor"].transform(X),
        cluster_labels
    )
    silhouette_scores[k] = score
    print(f"k={k}, silhouette score={score:.4f}")

# 최적 k 선택
best_k = max(silhouette_scores, key=silhouette_scores.get)
print(f"\n 최적의 클러스터 개수: k={best_k} (실루엣 계수={silhouette_scores[best_k]:.4f})")

# -------------------------------------
# 4. 최종 KMeans 실행 (best_k 사용)
# -------------------------------------
final_kmeans_pipe = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("cluster", KMeans(n_clusters=best_k, random_state=42, n_init=200))
])

df["cluster"] = final_kmeans_pipe.fit_predict(X)

# --- 스케일링된 데이터 삽입 ---
# 스케일러만 파이프라인에서 추출하여 데이터 변환
scaler = final_kmeans_pipe.named_steps["preprocessor"]
scaled_data = scaler.transform(X)
scaled_cols = [f"{col}_scaled" for col in num_vars]

# 스케일링된 데이터프레임을 생성하여 df_result에 합치기
df_scaled_vars = pd.DataFrame(
    scaled_data, 
    columns=scaled_cols, 
    index=X.index
)
df_concat = pd.concat([df, df_scaled_vars], axis=1)

# -------------------------------------
# 5. 군집별 프로파일 요약
# -------------------------------------
cluster_profile = df_concat.groupby("cluster")[scaled_cols].mean()

out_file1 = "segment/02_1_Profile_Kmeans_clustering_with_silhouette.csv"
out_file2 = "segment/03_1_Classified_Kmeans_clustering_with_silhouette.csv"
cluster_profile.to_csv(out_file1, encoding='cp949')
print("\n군집별 평균 카테고리 점수 저장완료 :", out_file1)

df_concat.to_csv(out_file2, index=False, encoding='cp949')
print("설문조사 세그먼트 입력 저장완료 :", out_file2)
print("\n[군집별 평균 카테고리 점수]")

# 클러스터 프로필의 수 확인
cluster_counts = df_concat['cluster'].value_counts().sort_index()
cluster_profile['count'] = cluster_counts
cluster_profile

# 2번 세그먼트는 소극적 관망자 형으로 고객으로는 부적합
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, OneHotEncoder, MinMaxScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# 0. 데이터 로드
file_path = "segment/01_category_scores_with_demo_multitag_normalized.csv"
df = pd.read_csv(file_path, encoding="cp949")

# 1. 입력 데이터 (df_result = 인구통계 + 카테고리 점수 데이터)
id_col = "id"
demo_cols = ["SQ3", "HQ2","SQ4", "SQ5", "SQ8", "DQ1", "DQ3", "DQ3_1", "DQ4"]

# 범주형/수치형 구분
cat_vars = demo_cols
num_vars = [c for c in df_result.columns if c not in demo_cols + [id_col]]

X = df_result.drop(columns=[id_col])
X_nums = df_result.drop(columns=[id_col]+ cat_vars)

# 2. 전처리 파이프라인
preprocessor = ColumnTransformer(
    transformers=[
        ("num", MinMaxScaler(), num_vars), 
        # ("num", "passthrough", num_vars),  # 그대로 사용
        ("cat", OneHotEncoder(handle_unknown="ignore"), cat_vars)
    ]
)


# 3. 최적 클러스터 개수 탐색 (실루엣 계수)
silhouette_scores = {}
X_preprocessed = preprocessor.fit_transform(X)

for k in range(4, 9):  # 4~8개 클러스터 탐색
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=100)
    cluster_labels = kmeans.fit_predict(X_preprocessed)
    
    # 실루엣 점수 계산
    score = silhouette_score(X_preprocessed, cluster_labels)
    silhouette_scores[k] = score
    print(f"k={k}, silhouette score={score:.4f}")

# 최적 k 선택
best_k = max(silhouette_scores, key=silhouette_scores.get)
print(f"\n 최적의 클러스터 개수: k={best_k} (실루엣 계수={silhouette_scores[best_k]:.4f})")


# 4. 최종 KMeans 실행 (best_k 사용)
final_kmeans = KMeans(n_clusters=best_k, random_state=42, n_init=100)
cluster_labels = final_kmeans.fit_predict(X_preprocessed)

df_result["cluster"] = cluster_labels

# --- 스케일링된 데이터 삽입 ---
# ColumnTransformer의 'num' 변환기만 별도로 사용하여 수치형 데이터를 스케일링
num_preprocessor = preprocessor.named_transformers_['num']
scaled_num_data = num_preprocessor.transform(X[num_vars])

scaled_cols = [f"{col}_scaled" for col in num_vars]
df_scaled_vars = pd.DataFrame(scaled_num_data, columns=scaled_cols, index=df_result.index)

df_concat = pd.concat([df_result, df_scaled_vars], axis=1)

# -------------------------------------
# 5. 군집별 프로파일 요약
# -------------------------------------
cluster_profile = df_concat.groupby("cluster")[num_vars].mean()
out_file1 = "segment/02_9_Profile_Kmeans_clustering_with_silhouette.csv"
out_file2 = "segment/03_9_Classified_Kmeans_clustering_with_silhouette.csv"
cluster_profile.to_csv(out_file1, index=False, encoding= 'cp949')
print("\n군집별 평균 카테고리 점수 저장완료 :", out_file1)

df_concat.to_csv(out_file2, index=False, encoding= 'cp949')
print("\n설문조사 세그먼트 입력 저장완료 :", out_file2)
print("\n[군집별 평균 카테고리 점수]")

# 클러스터 프로필의 수 확인
cluster_counts = df_result['cluster'].value_counts().sort_index()
cluster_profile['count'] = cluster_counts
cluster_profile
```

### 번외 - 다양한 군집화 알고리즘을 이용한 세그먼트
- KMeans, GMM, Hierarchical, DBSCAN 네 가지를 비교

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score


# 1. 데이터 로드
df = pd.read_csv("segment/01_category_scores_with_demo_multitag_normalized.csv", encoding="cp949")

id_col = "id"  # 실제 ID 컬럼명 맞게 수정
demo_cols = ["SQ3", "HQ2","SQ4", "SQ5", "SQ8", "DQ1", "DQ3", "DQ3_1", "DQ4"]

num_vars = [c for c in df.columns if c not in demo_cols + [id_col]]
X = df[num_vars].copy()

# MinMax 스케일링 (기본 스케일러 선택)
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# 2. KMeans
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
labels_kmeans = kmeans.fit_predict(X_scaled)
score_kmeans = silhouette_score(X_scaled, labels_kmeans)
print(f"KMeans silhouette (k=4): {score_kmeans:.4f}")

# -----------------------------
# 3. Gaussian Mixture Model (GMM)
# -----------------------------
gmm = GaussianMixture(n_components=4, random_state=42, n_init=10)
labels_gmm = gmm.fit_predict(X_scaled)
score_gmm = silhouette_score(X_scaled, labels_gmm)
print(f"GMM silhouette (k=4): {score_gmm:.4f}")

# -----------------------------
# 4. Hierarchical (Agglomerative Clustering)
# -----------------------------
hc = AgglomerativeClustering(n_clusters=4)
labels_hc = hc.fit_predict(X_scaled)
score_hc = silhouette_score(X_scaled, labels_hc)
print(f"Hierarchical silhouette (k=4): {score_hc:.4f}")

# -----------------------------
# 5. DBSCAN
# -----------------------------
dbscan = DBSCAN(eps=0.5, min_samples=10)
labels_db = dbscan.fit_predict(X_scaled)

# DBSCAN은 일부 점을 -1 (노이즈)로 할당하므로, 모든 값이 같은 군집이면 silhouette 불가
if len(set(labels_db)) > 1 and -1 not in set(labels_db):
    score_db = silhouette_score(X_scaled, labels_db)
    print(f"DBSCAN silhouette: {score_db:.4f}")
else:
    print(f"DBSCAN 결과: {len(set(labels_db))}개 클러스터 (노이즈 포함) → 실루엣 점수 계산 불가")

# -----------------------------
# 6. 결과 저장
# -----------------------------
df["cluster_kmeans"] = labels_kmeans
df["cluster_gmm"] = labels_gmm
df["cluster_hc"] = labels_hc
df["cluster_dbscan"] = labels_db

df.to_csv("segment/99_clustering_comparison_results.csv", index=False, encoding="cp949")
print("\n 군집 라벨 저장 완료: segment/03_clustering_comparison_results.csv")

cluster_profile = df.groupby("cluster_kmeans")[num_vars].count()
```

- **DBSCAN이 압도적으로 높은 점수를 낸 건, 클러스터링은 2개까지만 되면서, 고객 특성을 단순화 한것뿐으로 확인됨**

```python
# 통계학적 비율을 기반으로 300~500명 사이로 축소

import pandas as pd

# 1. 데이터 불러오기
file_path = "segment/03_2_Classified_Kmeans_clustering_불필요한세그먼트삭제.csv"
df = pd.read_csv(file_path, encoding="cp949")

# 2. 인구비율 조정
## 60대 이상 절반 축소
df_60 = df[df["HQ2"] == "60대 이상"]
df_not60 = df[df["HQ2"] != "60대 이상"]

df_60_reduced = df_60.sample(frac=0.5, random_state=42)
df_reduced = pd.concat([df_not60, df_60_reduced]).reset_index(drop=True)

## 20대 층화추출 비율 조정

df_20 = df_reduced[df_reduced["HQ2"] == "20대"]
df_reduced = df_reduced[df_reduced["HQ2"] != "20대"]

# -------------------------------
# 3. 층화표본추출 (연령 × 성별 × 가구형태 기준)
# -------------------------------
strata_cols = ["HQ2", "SQ3", "DQ1", "cluster"]

# 전체의 20% 정도 샘플링 (≈ 300~500명)
sample_fraction = 0.20

sample_df_not20 = (
    df_reduced.groupby(strata_cols, group_keys=False)
    .apply(lambda x: x.sample(frac=sample_fraction, random_state=42))
    .reset_index(drop=True)
)

sample_fraction = 0.50

sample_df_20 = (
    df_20.groupby(strata_cols, group_keys=False)
    .apply(lambda x: x.sample(frac=sample_fraction, random_state=42))
    .reset_index(drop=True)
)

sample_df = pd.concat([sample_df_not20, sample_df_20]).reset_index(drop=True)

# -------------------------------
# 4. 결과 확인
# -------------------------------
print("축소 전 데이터 크기 :", df.shape)
print("60대 절반 축소 후 크기 :", df_reduced.shape)
print("최종 샘플 크기 :", sample_df.shape)

print("\n샘플 분포 예시 (상위 10개):")
print(sample_df[strata_cols].value_counts(normalize=True).head(10))

# -------------------------------
# 5. 결과 저장
# -------------------------------
sample_df.to_csv("segment/04_persona_sample_stratified.csv", index=False, encoding="cp949")
```

```python
# 초기 설정
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 설정 (Mac과 Windows에 따라 다를 수 있습니다)
# Windows
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False # 마이너스 기호 깨짐 방지
```

## 1) 연령/성별 분포

```python
# 05_Persona_info.csv 파일 불러오기
df = pd.read_csv('segment/05_Persona_info.csv', encoding='cp949')

# 연령대 그룹 순서 지정
age_order = ['20대', '30대', '40대', '50대', '60대 이상']

# 연령대별 페르소나 비율 계산
# age_group을 인덱스로, label을 컬럼으로 하여 비율 계산
age_persona_distribution = df.groupby('age')['label'].value_counts(normalize=True).mul(100).unstack(fill_value=0)

# 인덱스 순서 재정렬
age_persona_distribution = age_persona_distribution.reindex(index=age_order, fill_value=0)

fig, axes = plt.subplots(1, 2, figsize=(18, 5))
fig.suptitle('페르소나별 연령 및 성별 분포', fontsize=20)

# 연령대 분포 데이터 시각화
age_persona_distribution.plot(kind='bar', ax=axes[0], stacked=True, colormap='tab20c')
axes[0].set_title('연령대별 페르소나 분포 (비율)', fontsize=18)
axes[0].set_xlabel('연령대', fontsize=14)
axes[0].set_ylabel('비율 (%)', fontsize=12)
axes[0].tick_params(axis='x', rotation=0, labelsize=12)  # X축 라벨 회전 방지
axes[0].legend(title='페르소나', bbox_to_anchor=(1.05, 1), loc='upper left')

# 성별 분포 데이터 시각화
gender_distribution = df.groupby('label')['gender'].value_counts(normalize=True).mul(100).unstack(fill_value=0)
gender_distribution.plot(kind='bar', stacked=True, ax=axes[1], colormap='viridis')
axes[1].set_title('페르소나별 성별 분포 (비율)', fontsize=16)
axes[1].set_xlabel('')
axes[1].set_ylabel('비율 (%)', fontsize=12)
axes[1].tick_params(axis='x', rotation=0, labelsize=12 )
axes[1].legend(title='성별', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

## 2) 클러스터별 특징 점수 시각화

```python
import numpy as np
from math import pi

# 02_3_Profile_Kmeans_clustering_with_silhouette_labeled.csv 파일 불러오기
df = pd.read_csv('segment/02_3_Profile_Kmeans_clustering_with_silhouette_labeled.csv', encoding='cp949')

# **문제 해결: 컬럼 이름에서 작은따옴표 제거**
df.columns = df.columns.str.replace("'", "")

# 레이더 차트에 사용할 특성 점수 열 선택
features = [
    'brand_loyalty_scaled', 
    'cooking_convenience_scaled', 
    'health_orientation_scaled', 
    'hmr_preference_scaled', 
    'premium_orientation_scaled', 
    'price_sensitivity_scaled', 
    'variety_seeking_scaled'
]

# 각 특징 점수 이름에 한글 라벨 추가
feature_labels = [
    '브랜드 충성도', '편의성 추구', '건강 지향성', 'HMR 선호도', 
    '프리미엄 지향성', '가격 민감도', '다양성 추구'
]

def make_radar_chart(ax, data, label):
    num_vars = len(feature_labels)
    angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
    angles += angles[:1]

    values = data.values.flatten().tolist()
    values += values[:1]

    ax.plot(angles, values, linewidth=2, linestyle='solid', label=label)
    ax.fill(angles, values, alpha=0.25)
    
    ax.set_yticklabels([])
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(feature_labels, fontsize=12)
 
    ax.set_ylim(0, 1)

# 전체 페르소나 목록
personas = df['label'].unique()

# **3. 차트 그리기**: 1행 4열 구조
fig, axes = plt.subplots(nrows=1, ncols=4, figsize=(18, 5), subplot_kw={'polar': True})
axes = axes.flatten()

for i, persona in enumerate(personas):
    data = df[df['label'] == persona][features]
    ax = axes[i]
    make_radar_chart(ax, data, persona)
    ax.set_title(persona, size=16, color='black', pad=20)
    
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.suptitle('페르소나별 특징 점수 레이더 차트', fontsize=20, y=1.0)
plt.show()
```
